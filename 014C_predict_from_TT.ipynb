```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from astropy.io import fits
from astropy.table import Table

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

### Load data

```python
##### DATA FOR TRAINING #####
with fits.open("../data/20260731/20260731_RT1.fits") as hdul:
    RT1 = Table(hdul[1].data).to_pandas()

with fits.open("../data/20260731/20260731_RT2.fits") as hdul:
    TT = Table(hdul[1].data).to_pandas()

print(len(RT1))
print(len(TT))
print("")

##### DATA FOR TESTING #####

with fits.open("../data/20260807/20260807_RT1.fits") as hdul:
    rt1 = Table(hdul[1].data).to_pandas()

with fits.open("../data/20260807/20260807_RT2.fits") as hdul:
    tt = Table(hdul[1].data).to_pandas()

print(len(rt1))
print(len(tt))
```

### Keep useful columns

```python
bbc_channels = [
"BBC02u",
"BBC03u",
"BBC04u",
"BBC05u",
"BBC06u",
"BBC10u",
"BBC11u",
"BBC13u",
"BBC14u"
]
```

```python
# training data
RT1_keep = RT1[["JD", "RIGHT_POL", "LEFT_POL"]].copy()
TT_keep = TT[["JD", "RIGHT_POL", "LEFT_POL"] +bbc_channels].copy()

# training data
rt1_keep = rt1[["JD", "RIGHT_POL", "LEFT_POL"]].copy()
tt_keep = tt[["JD", "RIGHT_POL", "LEFT_POL"] +bbc_channels].copy()
```

### sort by JD (probably useless but just in case)

```python
#training data
RT1_keep = RT1_keep.sort_values("JD")
TT_keep = TT_keep.sort_values("JD")
#testing data
rt1_keep = rt1_keep.sort_values("JD")
tt_keep = tt_keep.sort_values("JD")
```

### Synchronise by JD

```python
data = pd.merge_asof(
    RT1_keep,
    TT_keep,
    on="JD",
    direction="nearest",
    suffixes=("_RT1", "_TT")
)

print(data.shape)

data_test = pd.merge_asof(
    rt1_keep,
    tt_keep,
    on="JD",
    direction="nearest",
    suffixes=("_RT1", "_TT")
)

print(data_test.shape)
```

### check synchronisation, no gain

```python
#training data
plt.figure(figsize=(14,5))

plt.plot(
    data["RIGHT_POL_RT1"],
    label="RT1"
)

plt.plot(
    data["RIGHT_POL_TT"],
    label="TT"
)

plt.legend()
plt.grid()
#plt.xlim(90000,100000)
plt.title("training data")
plt.show()

#test data
plt.figure(figsize=(14,5))

plt.plot(
    data_test["RIGHT_POL_RT1"],
    label="RT1"
)

plt.plot(
    data_test["RIGHT_POL_TT"],
    label="TT"
)

plt.legend()
plt.grid()
#plt.xlim(90000,100000)
plt.title("testing data")
plt.show()
```

### Calibrate TT using the Sun

```python
sun_mask = (
    data["RIGHT_POL_RT1"]
    >
    data["RIGHT_POL_RT1"].median()
    + 3*data["RIGHT_POL_RT1"].std()
)

sun_mask_test = (
    data_test["RIGHT_POL_RT1"]
    >
    data_test["RIGHT_POL_RT1"].median()
    + 3*data_test["RIGHT_POL_RT1"].std()
)
```

```python
# Calibrate Right polarization
#training data
gain_R = (
    data.loc[sun_mask,"RIGHT_POL_RT1"].mean()
    /
    data.loc[sun_mask,"RIGHT_POL_TT"].mean()
)

data["RIGHT_POL_TT"] *= gain_R

#test data
gain_R_test = (
    data_test.loc[sun_mask_test,"RIGHT_POL_RT1"].mean()
    /
    data_test.loc[sun_mask_test,"RIGHT_POL_TT"].mean()
)

data_test["RIGHT_POL_TT"] *= gain_R_test
```

```python
# Calibrate left polarization
#training data
gain_L = (
    data.loc[sun_mask,"LEFT_POL_RT1"].mean()
    /
    data.loc[sun_mask,"LEFT_POL_TT"].mean()
)

data["LEFT_POL_TT"] *= gain_L

#test data
gain_L_test = (
    data_test.loc[sun_mask_test,"LEFT_POL_RT1"].mean()
    /
    data.loc[sun_mask_test,"LEFT_POL_TT"].mean()
)

data_test["LEFT_POL_TT"] *= gain_L_test

```

```python
# Normalise all channels
#training data
RT1_N = pd.DataFrame(index=data.index)
TT_N  = pd.DataFrame(index=data.index)

#test data
RT1_N_test = pd.DataFrame(index=data_test.index)
TT_N_test  = pd.DataFrame(index=data_test.index)
```

```python
# Right and Left
#training data
RT1_N["RIGHT_POL"] = (
    data["RIGHT_POL_RT1"]
    -
    data["RIGHT_POL_RT1"].median()
) / data["RIGHT_POL_RT1"].std()

RT1_N["LEFT_POL"] = (
    data["LEFT_POL_RT1"]
    -
    data["LEFT_POL_RT1"].median()
) / data["LEFT_POL_RT1"].std()

TT_N["RIGHT_POL"] = (
    data["RIGHT_POL_TT"]
    -
    data["RIGHT_POL_TT"].median()
) / data["RIGHT_POL_TT"].std()

TT_N["LEFT_POL"] = (
    data["LEFT_POL_TT"]
    -
    data["LEFT_POL_TT"].median()
) / data["LEFT_POL_TT"].std()

########################
#test data
RT1_N_test["RIGHT_POL"] = (
    data_test["RIGHT_POL_RT1"]
    -
    data_test["RIGHT_POL_RT1"].median()
) / data_test["RIGHT_POL_RT1"].std()

RT1_N_test["LEFT_POL"] = (
    data_test["LEFT_POL_RT1"]
    -
    data_test["LEFT_POL_RT1"].median()
) / data_test["LEFT_POL_RT1"].std()

TT_N_test["RIGHT_POL"] = (
    data_test["RIGHT_POL_TT"]
    -
    data_test["RIGHT_POL_TT"].median()
) / data_test["RIGHT_POL_TT"].std()

TT_N_test["LEFT_POL"] = (
    data_test["LEFT_POL_TT"]
    -
    data_test["LEFT_POL_TT"].median()
) / data_test["LEFT_POL_TT"].std()
```

```python
# BBC Channels
#training data
for ch in bbc_channels:

    TT_N[ch] = (
        data[ch]
        -
        data[ch].median()
    ) / data[ch].std()

#test data
for ch in bbc_channels:
    TT_N_test[ch] = (
        data_test[ch]
        -
        data_test[ch].median()
    ) / data_test[ch].std()
```

### Check Solar peak matching

```python
#training data
plt.figure(figsize=(14,5))
plt.plot(
    TT_N["RIGHT_POL"],
    label="TT"
)
plt.plot(
    RT1_N["RIGHT_POL"],
    label="RT1"
)
plt.legend()
plt.grid()
plt.title("Training data: after calibration and normalization")
#plt.xlim(140000,150000)
plt.show()

#test data
plt.figure(figsize=(14,5))
plt.plot(
    TT_N_test["RIGHT_POL"],
    label="TT"
)
plt.plot(
    RT1_N_test["RIGHT_POL"],
    label="RT1"
)
plt.legend()
plt.grid()
plt.title("Test data: after calibration and normalization")
#plt.xlim(140000,150000)
plt.show()
```

### Polarization asymmetry

```python
#training data
RT1_N["POL_ASYM"] = (RT1_N["RIGHT_POL"]-RT1_N["LEFT_POL"]) / (RT1_N["RIGHT_POL"]+RT1_N["LEFT_POL"]+ 1e-6)
#test data
RT1_N_test["POL_ASYM"] = (RT1_N_test["RIGHT_POL"]-RT1_N_test["LEFT_POL"]) / (RT1_N_test["RIGHT_POL"]+RT1_N_test["LEFT_POL"]+ 1e-6)
```

```python
#training data
TT_N["POL_ASYM"] = (TT_N["RIGHT_POL"]-TT_N["LEFT_POL"]) / (TT_N["RIGHT_POL"]+TT_N["LEFT_POL"]+ 1e-6)
#test data
TT_N_test["POL_ASYM"] = (TT_N_test["RIGHT_POL"]-TT_N_test["LEFT_POL"]) / (TT_N_test["RIGHT_POL"]+TT_N_test["LEFT_POL"]+ 1e-6)
```

### Time Series Settings

```python
# for training 
window = 20
feature_rows = []
target_rows = []
target_index = []
# for testing
window_test = 20
feature_rows_test = []
target_rows_test = []
target_index_test = []
```

### Build ML dataset

```python
#prepare the training set 
for i in range(window, len(data)):

    row = []

    seg = TT_N["RIGHT_POL"].iloc[i-window:i]
    row.append(seg.mean())
    row.append(seg.std())

    seg = TT_N["LEFT_POL"].iloc[i-window:i]
    row.append(seg.mean())
    row.append(seg.std())

    seg = TT_N["POL_ASYM"].iloc[i-window:i]
    row.append(seg.mean())
    row.append(seg.std())

    for ch in bbc_channels:

        seg = TT_N[ch].iloc[i-window:i]

        row.append(seg.mean())
        row.append(seg.std())

    feature_rows.append(row)

    target_rows.append(
        RT1_N["RIGHT_POL"].iloc[i]
    )

    target_index.append(i)
```

```python
#prepare the test set
for i in range(window_test, len(data_test)):

    row_test = []

    seg = TT_N_test["RIGHT_POL"].iloc[i-window_test:i]
    row_test.append(seg.mean())
    row_test.append(seg.std())

    seg = TT_N_test["LEFT_POL"].iloc[i-window_test:i]
    row_test.append(seg.mean())
    row_test.append(seg.std())

    seg = TT_N_test["POL_ASYM"].iloc[i-window_test:i]
    row_test.append(seg.mean())
    row_test.append(seg.std())

    for ch in bbc_channels:
        seg = TT_N_test[ch].iloc[i-window_test:i]
        row_test.append(seg.mean())
        row_test.append(seg.std())

    feature_rows_test.append(row_test)

    target_rows_test.append(
        RT1_N_test["RIGHT_POL"].iloc[i]
    )

    target_index_test.append(i)
```

### Arrays

```python
X = np.array(feature_rows)
y = np.array(target_rows)

print(X.shape)
print(y.shape)
print("")

Xt = np.array(feature_rows_test)
yt = np.array(target_rows_test)

print(Xt.shape)
print(yt.shape)
```

### Train/Test split

```python
#split = int(
#    0.8 * len(X)
#)
#
#X_train = X[:split]
#X_test  = X[split:]
#
#y_train = y[:split]
#y_test  = y[split:]
#
#index_test = np.array(
#    target_index[split:]
#)
```

```python
X_train = X
y_train = y

X_test = Xt
y_test = yt

index_test = np.array(target_index_test)
```

### Train model

```python
model = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

model.fit(
    X_train,
    y_train
)
```

### Predict

```python
pred = model.predict(
    X_test
)
```

### Observed vs Predicted

```python
plt.figure(figsize=(14,5))

plt.plot(
    y_test,
    label="Observed RT1"
)

plt.plot(
    pred,
    label="Predicted from TT"
)

plt.legend()
plt.grid()

plt.show()
```

### $R^2$

```python
print(
    "R² =",
    r2_score(
        y_test,
        pred
    )
)
```

### Residual

```python
residual = (
    y_test
    -
    pred
)
```

```python
plt.figure(figsize=(14,5))

plt.plot(residual)

plt.axhline(
    0,
    color="black"
)

plt.grid()

plt.show()
```

### Detect anomalies

```python
sigma = residual.std()

anomaly = (
    np.abs(residual)
    >
    1.5*sigma
)

print(
    "Number of anomalies:",
    anomaly.sum()
)
```

### Plot anomalies on RT1

```python
plt.figure(figsize=(14,5))

plt.plot(
    RT1_N_test["RIGHT_POL"],
    color="steelblue",
    label="RT1"
)

plt.scatter(index_test[anomaly],
    RT1_N_test["RIGHT_POL"]
    .iloc[index_test[anomaly]],

    color="red",
    s=30,

    label="Anomaly"
)

plt.legend()
plt.grid()

plt.title(
    "Times when RT1 differs from TT prediction"
)

plt.show()
```
