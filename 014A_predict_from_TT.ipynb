```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from astropy.io import fits
from astropy.table import Table

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

### Load data

```python
with fits.open("../data/20261010_taitam/20261010_RT1.fits") as hdul:
    RT1 = Table(hdul[1].data).to_pandas()

with fits.open("../data/20261010_taitam/20261010_TAITAM.fits") as hdul:
    TT = Table(hdul[1].data).to_pandas()

print(len(RT1))
print(len(TT))
```

### Keep useful columns

```python
bbc_channels = [
"BBC02u",
"BBC03u",
"BBC04u",
"BBC05u",
"BBC06u",
"BBC10u",
"BBC11u",
"BBC13u",
"BBC14u"
]
```

```python
RT1_keep = RT1[
    ["JD", "RIGHT_POL", "LEFT_POL"]
].copy()

TT_keep = TT[
    ["JD", "RIGHT_POL", "LEFT_POL"] +
    bbc_channels
].copy()
```

### sort by JD (probably useless by just in case)

```python
RT1_keep = RT1_keep.sort_values("JD")
TT_keep = TT_keep.sort_values("JD")
```

### Synchronise by JD

```python
data = pd.merge_asof(
    RT1_keep,
    TT_keep,
    on="JD",
    direction="nearest",
    suffixes=("_RT1", "_TT")
)

print(data.shape)
```

### check synchronisation, no gain

```python
plt.figure(figsize=(14,5))

plt.plot(
    150*data["RIGHT_POL_RT1"],
    label="RT1"
)

plt.plot(
    data["RIGHT_POL_TT"]+6000,
    label="TT"
)

plt.legend()
plt.grid()

plt.show()
```

### Calibrate TT using the Sun

```python
sun_mask = (
    data["RIGHT_POL_RT1"]
    >
    data["RIGHT_POL_RT1"].median()
    + 3*data["RIGHT_POL_RT1"].std()
)
```

```python
# Calibrate Right polarization
gain_R = (
    data.loc[sun_mask,"RIGHT_POL_RT1"].mean()
    /
    data.loc[sun_mask,"RIGHT_POL_TT"].mean()
)

data["RIGHT_POL_TT"] *= gain_R
```

```python
# Calibrate left polarization
gain_L = (
    data.loc[sun_mask,"LEFT_POL_RT1"].mean()
    /
    data.loc[sun_mask,"LEFT_POL_TT"].mean()
)

data["LEFT_POL_TT"] *= gain_L
```

```python
# Normalise all channels
RT1_N = pd.DataFrame(index=data.index)
TT_N  = pd.DataFrame(index=data.index)
```

```python
# Right and Left
RT1_N["RIGHT_POL"] = (
    data["RIGHT_POL_RT1"]
    -
    data["RIGHT_POL_RT1"].median()
) / data["RIGHT_POL_RT1"].std()

RT1_N["LEFT_POL"] = (
    data["LEFT_POL_RT1"]
    -
    data["LEFT_POL_RT1"].median()
) / data["LEFT_POL_RT1"].std()

TT_N["RIGHT_POL"] = (
    data["RIGHT_POL_TT"]
    -
    data["RIGHT_POL_TT"].median()
) / data["RIGHT_POL_TT"].std()

TT_N["LEFT_POL"] = (
    data["LEFT_POL_TT"]
    -
    data["LEFT_POL_TT"].median()
) / data["LEFT_POL_TT"].std()
```

```python
# BBC Channels
for ch in bbc_channels:

    TT_N[ch] = (
        data[ch]
        -
        data[ch].median()
    ) / data[ch].std()
```

### Check Solar peak matching

```python
plt.figure(figsize=(14,5))

plt.plot(
    TT_N["RIGHT_POL"],
    label="TT"
)

plt.plot(
    RT1_N["RIGHT_POL"],
    label="RT1"
)

plt.legend()
plt.grid()

plt.title(
    "After calibration and normalization"
)

plt.show()
```

### Polarization asymmetry

```python
RT1_N["POL_ASYM"] = (

    RT1_N["RIGHT_POL"]
    -
    RT1_N["LEFT_POL"]

) / (

    RT1_N["RIGHT_POL"]
    +
    RT1_N["LEFT_POL"]
    + 1e-6
)
```

```python
TT_N["POL_ASYM"] = (

    TT_N["RIGHT_POL"]
    -
    TT_N["LEFT_POL"]

) / (

    TT_N["RIGHT_POL"]
    +
    TT_N["LEFT_POL"]
    + 1e-6
)
```

### Time Series Settings

```python
window = 20

feature_rows = []

target_rows = []

target_index = []
```

### Build ML dataset

```python
for i in range(window, len(data)):

    row = []

    seg = TT_N["RIGHT_POL"].iloc[i-window:i]
    row.append(seg.mean())
    row.append(seg.std())

    seg = TT_N["LEFT_POL"].iloc[i-window:i]
    row.append(seg.mean())
    row.append(seg.std())

    seg = TT_N["POL_ASYM"].iloc[i-window:i]
    row.append(seg.mean())
    row.append(seg.std())

    for ch in bbc_channels:

        seg = TT_N[ch].iloc[i-window:i]

        row.append(seg.mean())
        row.append(seg.std())

    feature_rows.append(row)

    target_rows.append(
        RT1_N["RIGHT_POL"].iloc[i]
    )

    target_index.append(i)
```

### Arrays

```python
X = np.array(feature_rows)

y = np.array(target_rows)

print(X.shape)
print(y.shape)
```

### Train/Test split

```python
split = int(
    0.8 * len(X)
)

X_train = X[:split]
X_test  = X[split:]

y_train = y[:split]
y_test  = y[split:]

index_test = np.array(
    target_index[split:]
)
```

### Train model

```python
model = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

model.fit(
    X_train,
    y_train
)
```

### Predict

```python
pred = model.predict(
    X_test
)
```

### Observed vs Predicted

```python
plt.figure(figsize=(14,5))

plt.plot(
    y_test,
    label="Observed RT1"
)

plt.plot(
    pred,
    label="Predicted from TT"
)

plt.legend()
plt.grid()

plt.show()
```

### $R^2$

```python
print(
    "R² =",
    r2_score(
        y_test,
        pred
    )
)
```

### Residual

```python
residual = (
    y_test
    -
    pred
)
```

```python
plt.figure(figsize=(14,5))

plt.plot(residual)

plt.axhline(
    0,
    color="black"
)

plt.grid()

plt.show()
```

### Detect anomalies

```python
sigma = residual.std()

anomaly = (
    np.abs(residual)
    >
    3*sigma
)

print(
    "Number of anomalies:",
    anomaly.sum()
)
```

### Plot anomalies on RT1

```python
plt.figure(figsize=(14,5))

plt.plot(
    RT1_N["RIGHT_POL"],
    color="steelblue",
    label="RT1"
)

plt.scatter(
    index_test[anomaly],

    RT1_N["RIGHT_POL"]
    .iloc[index_test[anomaly]],

    color="red",
    s=30,

    label="Anomaly"
)

plt.legend()
plt.grid()

plt.title(
    "Times when RT1 differs from TT prediction"
)

plt.show()
```
